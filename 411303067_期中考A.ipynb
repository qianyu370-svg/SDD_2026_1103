````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (2) 建立資料庫與初始資料
# 每次執行都先刪除舊資料表，重建 activities 與 registrations 表，並寫入初始資料

def init_db():
    conn = sqlite3.connect('club_events.db')
    cursor = conn.cursor()
    try:
        # 刪除舊表確保每次執行都是乾淨的初始狀態
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立 activities 資料表
        cursor.execute("""
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立 registrations 資料表
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_name TEXT NOT NULL,
                count INTEGER NOT NULL,
                status TEXT NOT NULL
            )
        """)

        # 放入初始活動資料
        initial_activities = [
            ('迎新晚會', '2026-11-20', 6),
            ('程式工作坊', '2026-11-25', 4),
            ('籃球友誼賽', '2026-11-28', 3)
        ]
        cursor.executemany("""
            INSERT INTO activities (name, date, capacity)
            VALUES (?, ?, ?)
        """, initial_activities)

        conn.commit()
    finally:
        conn.close()

# 執行初始化
init_db()
print("資料庫建立成功，並已載入初始資料！")
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (3) 顯示資料表
# 透過 pandas 查詢並印出目前資料庫中的 activities 與 registrations 內容，方便隨時檢查狀態

def show_tables():
    conn = sqlite3.connect('club_events.db')
    try:
        print("=== 【 activities 活動表 】 ===")
        df_activities = pd.read_sql("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n=== 【 registrations 報名紀錄表 】 ===")
        df_registrations = pd.read_sql("SELECT * FROM registrations", conn)
        display(df_registrations)
    finally:
        conn.close()

# 執行顯示目前資料表
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (4) AI 解讀函式
# 呼叫 Groq 模型解析使用者的一句話，具備 429 錯誤重試機制、暫存防重複呼叫、嚴格白名單與正規表示式過濾

ai_cache = {}  # 暫存已呼叫過的對話，避免重複呼叫 AI

def parse_with_ai(user_input):
    # 若該句話已經查詢過，直接從暫存回傳
    if user_input in ai_cache:
        return ai_cache[user_input]

    prompt = f"""
請幫我分析下面使用者的一句話，並嚴格按照格式輸出：`動作,活動,人數`
動作只能從這裡選：報名, 取消, 查詢, 統計
活動只能從這裡選：迎新晚會, 程式工作坊, 籃球友誼賽, 各活動報名狀況, None
人數規則：若有提到人數則填數字，沒提到預設為 1；如果是查詢、取消、統計則輸出 0。
如果無法判斷或內容無關，請固定輸出：ERROR,None,0

使用者說的話：{user_input}
"""

    response_text = ""
    # 嘗試呼叫 Groq API，遇到 429 過量錯誤時重試最多 3 次
    for attempt in range(3):
        try:
            time.sleep(2)  # 每次呼叫後暫停 2 秒
            completion = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            response_text = completion.choices[0].message.content.strip()
            break
        except Exception as e:
            if "429" in str(e) or "rate_limit" in str(e).lower():
                print(f"遇到 429 流量限制，等待 30 秒後重試 (第 {attempt+1} 次)...")
                time.sleep(30)
            else:
                print(f"AI 呼叫發生錯誤: {e}")
                response_text = "ERROR,None,0"
                break
    else:
        response_text = "ERROR,None,0"

    # 安全防護：不直接相信 AI 回覆，進行白名單與正規表示式過濾
    valid_actions = ["報名", "取消", "查詢", "統計"]
    valid_activities = ["迎新晚會", "程式工作坊", "籃球友誼賽", "各活動報名狀況"]

    parts = response_text.split(",")
    if len(parts) >= 3:
        action = parts[0].strip()
        activity = parts[1].strip()
        raw_count_str = parts[2].strip()

        # 檢查動作是否合法
        if action not in valid_actions:
            action = "ERROR"

        # 檢查活動是否合法
        if activity not in valid_activities and action != "ERROR":
            activity = "None"
            action = "ERROR"

        # 用正規表示式萃取人數數字
        match = re.search(r'\d+', raw_count_str)
        if match:
            count = int(match.group())
        else:
            count = 1 if action == "報名" else 0
    else:
        action, activity, count = "ERROR", "None", 0

    final_result = (action, activity, count, response_text)
    ai_cache[user_input] = final_result
    return final_result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (5) AI 解讀測試
# 執行 9 句測試案例（包含正常、邊界、異常），印出 AI 原始回覆、系統解讀與通過幾句

test_cases = [
    ("正常", "我想報名迎新晚會，我和兩個朋友共 3 人", "報名", "迎新晚會", 3),
    ("正常", "幫我報名程式工作坊 1 個人", "報名", "程式工作坊", 1),
    ("正常", "查詢籃球友誼賽名額", "查詢", "籃球友誼賽", 0),
    ("正常", "我要取消迎新晚會", "取消", "迎新晚會", 1),
    ("正常", "請顯示各活動報名狀況", "統計", "各活動報名狀況", 0),
    ("邊界", "報名籃球友誼賽 2 個人", "報名", "籃球友誼賽", 2),
    ("邊界", "幫我查程式工作坊", "查詢", "程式工作坊", 0),
    ("異常", "報名火星探險活動 1 個人", "ERROR", "None", 0),
    ("異常", "亂打一通的測試文字", "ERROR", "None", 0)
]

passed_count = 0
print("=== 【 AI 解讀測試開始 】 ===")
for i, (test_type, text, exp_action, exp_activity, exp_count) in enumerate(test_cases, 1):
    action, activity, count, raw_resp = parse_with_ai(text)

    # 比對結果
    is_pass = (action == exp_action and activity == exp_activity and (exp_count == "*" or count == exp_count))
    if is_pass:
        passed_count += 1
        status_str = "【通過】"
    else:
        status_str = "【失敗】"

    print(f"\n測項 {i} [{test_type}]: {status_str}")
    print(f"  使用者說的話 : {text}")
    print(f"  AI 原始回覆 : {raw_resp}")
    print(f"  系統解讀結果 : 動作={action}, 活動={activity}, 人數={count}")
    print(f"  預期結果     : 動作={exp_action}, 活動={exp_activity}, 人數={exp_count}")

print(f"\n================================")
print(f"測試完成！總共通過 {passed_count} / {len(test_cases)} 句。")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (6) 功能函式
# 實作報名、取消、查詢與統計的資料庫商業邏輯，嚴格落實例外檢查（人數範圍、個人上限、名額是否足夠、查無紀錄等）

def process_request(student_name, user_input):
    action, activity, count, _ = parse_with_ai(user_input)

    # 錯誤判斷
    if action == "ERROR":
        return "系統無法理解您的需求，請重新輸入（例如：我想報名迎新晚會 2 人）。"

    conn = sqlite3.connect('club_events.db')
    cursor = conn.cursor()
    try:
        # 1. 統計查詢功能
        if action == "統計":
            cursor.setItem = cursor.execute("""
                SELECT a.name, a.date, a.capacity,
                       COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) AS registered_count
                FROM activities a
                LEFT JOIN registrations r ON a.name = r.activity_name
                GROUP BY a.id, a.name, a.date, a.capacity
            """)
            rows = cursor.fetchall()
            result_str = "【各活動報名統計狀況】\n"
            for row in rows:
                name, date, capacity, reg_count = row
                remaining = capacity - reg_count
                result_str += f"- 活動：{name} | 日期：{date} | 總名額：{capacity} | 已報名：{reg_count} | 剩餘名額：{remaining}\n"
            return result_str.strip()

        # 檢查活動是否存在於 activities 表中
        cursor.execute("SELECT capacity, date FROM activities WHERE name = ?", (activity,))
        act_row = cursor.fetchone()
        if act_row is None:
            return f"查無此活動「{activity}」，請確認活動名稱。"
        capacity, date = act_row

        # 2. 查詢功能
        if action == "查詢":
            cursor.execute("""
                SELECT COALESCE(SUM(CASE WHEN status = '已報名' THEN count ELSE 0 END), 0)
                FROM registrations WHERE activity_name = ?
            """, (activity,))
            reg_row = cursor.fetchone()
            reg_count = reg_row[0] if reg_row else 0
            remaining = capacity - reg_count
            return f"活動名稱：{activity}\n活動日期：{date}\n總名額：{capacity} 人\n目前已報名：{reg_count} 人\n剩餘名額：{remaining} 人"

        # 3. 報名功能
        if action == "報名":
            # 例外 1：人數超出範圍 (1～3 人)
            if not (1 <= count <= 3):
                return "拒絕報名：單次報名人數必須介於 1 到 3 人之間。"

            # 檢查是否重複報名同一個活動（有效狀態下）
            cursor.execute("""
                SELECT id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            if cursor.fetchone() is not None:
                return f"拒絕報名：您({student_name})已經報名過「{activity}」了，不可重複報名。"

            # 例外 2：個人報名活動數量超標（最多同時報名 2 個活動）
            cursor.execute("""
                SELECT COUNT(DISTINCT activity_name) FROM registrations
                WHERE student_name = ? AND status = '已報名'
            """, (student_name,))
            active_events_count = cursor.fetchone()[0]
            if active_events_count >= 2:
                return "拒絕報名：每位學生最多同時報名 2 個活動。"

            # 檢查目前活動已報名總人數
            cursor.execute("""
                SELECT COALESCE(SUM(CASE WHEN status = '已報名' THEN count ELSE 0 END), 0)
                FROM registrations WHERE activity_name = ?
            """, (activity,))
            reg_count = cursor.fetchone()[0]
            remaining = capacity - reg_count

            # 檢查名額是否充足
            if count > remaining:
                return f"拒絕報名：名額不足！目前剩餘名額僅剩 {remaining} 人，您欲報名 {count} 人。"

            # 寫入資料庫
            cursor.execute("""
                INSERT INTO registrations (student_name, activity_name, count, status)
                VALUES (?, ?, ?, '已報名')
            """, (student_name, activity, count))
            conn.commit()

            new_remaining = remaining - count
            return f"報名成功！{student_name} 成功報名「{activity}」共 {count} 人。目前該活動剩餘名額：{new_remaining} 人。"

        # 4. 取消功能
        if action == "取消":
            # 檢查是否有該學生的有效報名紀錄
            cursor.execute("""
                SELECT id, count FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity))
            reg_record = cursor.fetchone()

            # 例外 3：查無報名紀錄（取消失敗）
            if reg_record is None:
                return f"取消失敗：查無此有效報名紀錄（學生：{student_name}，活動：{activity}）。"

            reg_id, reg_count = reg_record

            # 將狀態改為 '已取消'
            cursor.execute("""
                UPDATE registrations SET status = '已取消' WHERE id = ?
            """, (reg_id,))
            conn.commit()

            return f"取消報名成功！已為 {student_name} 取消「{activity}」的報名，釋出 {reg_count} 個名額。"

    finally:
        conn.close()
````

````python
# ============================================================
# 第 7 格：主流程與流程測試
# 提醒：流程測試 8～10 步；最後呼叫 show_tables()
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (7) 主流程與流程測試
# 每次執行前重新初始化資料庫，並依序執行 9 個流程測試步驟，最後顯示資料表狀態

init_db()  # 重建並初始化資料庫

workflow_steps = [
    (1, "王小明", "查詢迎新晚會名額"),
    (2, "王小明", "報名迎新晚會，3個人"),
    (3, "李小華", "報名迎新晚會，3個人"),
    (4, "張大山", "報名迎新晚會，1個人"),
    (5, "李小華", "報名程式工作坊，1個人"),
    (6, "王小明", "報名籃球友誼賽，1個人"),
    (7, "王小明", "報名程式工作坊，1個人"),
    (8, "李小華", "取消迎新晚會"),
    (9, "幹部", "請顯示各活動報名狀況")
]

print("=== 【 流程測試開始 】 ===")
for step_num, name, text in workflow_steps:
    print(f"\n--- 步驟 {step_num} ---")
    print(f"【使用者】{name}：\"{text}\"")
    response = process_request(name, text)
    print(f"【系統回應】\n{response}")

print("\n=== 【 流程測試結束，最終資料表狀態 】 ===")
show_tables()
````
